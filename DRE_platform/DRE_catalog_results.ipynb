```python
all_countries = []
s3_base = 's3://wbg-geography01/GEP/DRE_Atlas/SETTLEMENTS'
for country in os.listdir(in_folder):
    if os.path.isdir(os.path.join(in_folder, country)) and not country.startswith("."):
        try:
            cntry = pycountry.countries.get(alpha_3=country)
        except:
            print(f"Country not found: {country}")
            continue        
        # list the files in the country folder
        country_folder = os.path.join(in_folder, country)
        fgb_file = [x for x in os.listdir(country_folder) if x.endswith(".fgb")]        
        csv_file = [x for x in os.listdir(country_folder) if x.endswith(".csv")]
        
        # print(f"{country}: FGB {os.path.exists(os.path.join(country_folder, fgb_file[0]))}, CSV {os.path.exists(os.path.join(country_folder, csv_file[0]))}")
        # add the new dataset
        # Resource title, resoutrce description
        all_countries.append(["Dataset", f"{cntry.alpha_3} DRE exploration results",description_text.format(platform_title="DRE exploration platform", platform_url="TBD", country=cntry.name),""])
        # Add resources to the dataset
        # Resource type, title, summary, filename
        all_countries.append(["Resource", "Polygonal settlement extents with results", 
                              "FlatGeoBuf file with the mini-grid potential results as complete settlement polygons", 
                              f'{s3_base}/{cntry.alpha_3}/{fgb_file[0]}'])
        all_countries.append(["Resource", "Tabular settlements with Results", 
                              "CSV file with the mini-grid potential results, settlemenrt shapes are stored as centroid coordinates",
                              f'{s3_base}/{cntry.alpha_3}/{csv_file[0]}']) 
        all_countries.append(["Resource", "Results field definitions", "Excel file defining the fields in the results files along with original source information", 
                              f'{s3_base}/{cntry.alpha_3}/DRE_results_field_metadata.xlsx'])

all_countries = pd.DataFrame(all_countries, columns=["Type", "Title", "Description", "Filename"])

all_countries.to_csv(out_file, index=False)
```
